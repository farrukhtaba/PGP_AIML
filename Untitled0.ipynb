```python
from google.colab import drive
drive.mount('/content/drive')
file_path = "/content/drive/MyDrive/data1.xlsx"
https://docs.google.com/spreadsheets/d/1aB2Gn1Tf0z164ALx_lYxh72Wjj38I-Rv/edit?usp=drive_link&ouid=116720462426417351288&rtpof=true&sd=true
import pandas as pd

EXCEL_FILE = "data1.xlsx"
TABLE_NAME = "rxp_patient"
PRIMARY_KEY = "PATID"   # change if needed

df = pd.read_excel(EXCEL_FILE)

sql_statements = []

for _, row in df.iterrows():
    set_clauses = []

    for column in df.columns:
        if column == PRIMARY_KEY:
            continue

        value = row[column]

        if pd.isna(value):
            set_clauses.append(f"{column} = NULL")
        elif isinstance(value, str):
            value = value.replace("'", "''")  # escape quotes
            set_clauses.append(f"{column} = '{value}'")
        else:
            set_clauses.append(f"{column} = {value}")

    where_clause = f"{PRIMARY_KEY} = {row[PRIMARY_KEY]}"

    sql = f"""UPDATE {TABLE_NAME}
SET {", ".join(set_clauses)}
WHERE {where_clause};"""

    sql_statements.append(sql)

# Write to SQL file
with open("update_rxp_patient.sql", "w") as f:
    for stmt in sql_statements:
        f.write(stmt + "\n\n")

print("SQL file generated: update_rxp_patient.sql")
```
